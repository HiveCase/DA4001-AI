## Question 2

```python
prompt = ChatPromptTemplate.from_template("What is the capital of {country}")
chain = prompt | ChatGroq(model="llama-3.3-70b-versatile") | StrOutputParser()
```

```python
chain.invoke({"country":"Bharat"})
```

```python
chain.invoke("France")
```

```python
prompt2 = ChatPromptTemplate.from_template("What is the national sports of {country} and What is the population of {city}")
```

```python
chain2 = prompt2 | ChatGroq(model="llama-3.3-70b-versatile") | StrOutputParser()
```

```python
chain2.invoke({"country":"India", "city": "Mumbai"})
```

## Question 3

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough
```

```python
runnable = RunnableParallel(passed_through=RunnablePassthrough(), new_key=lambda x: x["num"] + 10)
print(runnable.invoke({"num":5}))
```

## Question 6

```python
from langchain_core.runnables import RunnableLambda
```

### Option A

```python
a = RunnableLambda(lambda x: x + "world")
b = RunnableLambda(lambda x: x + "!")
chain = a | b
print(chain.invoke(" "))
```

### Option B

```python
a = RunnableLambda(lambda x: x + "world")
b = RunnableLambda(lambda x: x + "!")
chain = a | b
print(chain.invoke(" "))
```

### Option C

```python
a = RunnableLambda(lambda x: x + " Hello")
b = RunnableLambda(lambda x: x + "world!")
chain = a | b
print(chain.invoke("Hello"))
```

### Option D

```python
a = RunnableLambda(lambda x: x + " world")
b = RunnableLambda(lambda x: x + "!")
chain = a | b
print(chain.invoke("Hello"))
```

## Question 7
